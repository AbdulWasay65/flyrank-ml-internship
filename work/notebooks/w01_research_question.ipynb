## 2. The question: decision, action, cost of a wrong call

The decision is which pages should be prioritized for content-refresh review. The main user of the output would be an SEO or content team deciding where to spend limited review and optimization time. The output would be a ranked list of pages with signals indicating which pages may deserve attention first. A reviewer could then inspect the highest-ranked pages and decide whether a refresh is appropriate. A wrong recommendation has an opportunity cost: a team could spend time refreshing a page that does not need it, while a page that would benefit from review may be missed or delayed. ML can help by combining several measurable signals consistently rather than relying only on a single hand-written rule.

```python
# This cell is for CODE
print(f"Pages available for prioritization: {len(df):,}")
print(f"Declining pages: {(df['trend_direction'] == 'down').sum():,}")
print(f"Declining rate: {(df['trend_direction'] == 'down').mean():.3f}")
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```
